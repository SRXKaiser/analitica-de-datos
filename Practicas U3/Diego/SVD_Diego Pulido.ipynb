```python
# Librerías necesarias
import pandas as pd
import numpy as np
from sklearn.decomposition import TruncatedSVD
```

### **Liga de Datos de la UCI "Restaurant & consumer data".**
### **Descarga el archivo "restaurant+consumer+data.zip" y de ahí necesitaremos los archivos "rating_final.csv" y "geoplaces2.csv".**

https://archive.ics.uci.edu/dataset/232/restaurant+consumer+data



```python
# Carga de los archivos proporcionados
data1 = pd.read_csv("rating_final.csv", sep=",")
data2 = pd.read_csv("geoplaces2.csv", sep=",", encoding="latin-1")
print("Dimensiones:", data1.shape, data2.shape)
```

## **Ejercicio - 1**

### **Explica cuál es el propósito de usar el argumento "latin-1" al cargar el segundo archivo.**

El argumento `encoding="latin-1"` indica a pandas que interprete el archivo usando la codificación Latin-1 (ISO-8859-1). Se utiliza porque el archivo contiene caracteres especiales, como acentos y la letra ñ. Así se evita que aparezcan errores de decodificación o caracteres incorrectos al leer los nombres y demás datos de los restaurantes.


```python
# Del primer archivo obtenemos una matriz con 3 evaluaciones de los restaurantes:
# general, comida y servicio.
# Las evaluaciones pueden ser 0, 1 o 2, siendo 0 la menor calificación y 2 la
# mayor calificación:

data1.head()
```

```python
# Del segundo archivo obtenemos información diversa de cada restaurante:

data2.head(2)
```

```python
# Seleccionamos las columnas necesarias de cada archivo
lista_data1 = ["userID", "placeID", "food_rating", "service_rating"]
lista_data2 = ["placeID", "name"]

data1a = data1[lista_data1].copy()
data2a = data2[lista_data2].copy()
```

## **Ejercicio - 2**

```python
# Unimos las evaluaciones con los nombres de los restaurantes mediante placeID
df_combinado = pd.merge(data1a, data2a, on="placeID", how="inner")

print(df_combinado.shape)
df_combinado.head()
```

## **Ejercicio - 3**

```python
# Matriz de utilidad: restaurantes en filas, usuarios en columnas y
# evaluación de comida en cada celda. Los valores faltantes se sustituyen por 0.
UtMx_food = df_combinado.pivot_table(
    values="food_rating", index="name", columns="userID", fill_value=0
)

print("Dimensión de la matriz de Utilidad sobre la evaluación de la comida:")
print("(restaurantes, usuarios) =", UtMx_food.shape)
UtMx_food.head()
```

## **Ejercicio - 4**

### **En la Factorización SVD la cantidad de valores singulares será menor o igual al menor valor de los renglones o columnas de la matriz de utilidad UtMx.**

### **La factorización SVD de una matriz $A$ tiene la forma:**

$A_{m\times n} = U_{m\times m}\Sigma_{m\times n}V_{n\times n}^T$

```python
# SVD completa: se calcula el máximo número posible de componentes.
U_food, s_food, Vt_food = np.linalg.svd(UtMx_food, full_matrices=False)
nc_food = min(UtMx_food.shape)

print("Total de valores singulares basados en evaluación de la comida:", nc_food)
```

```python
# SVD truncada con el máximo número de componentes calculado anteriormente
SVD_food = TruncatedSVD(n_components=nc_food, random_state=1)
SVD_food.fit_transform(UtMx_food)

# Número mínimo de componentes cuya varianza acumulada alcanza al menos 90%
var_acum_food = np.cumsum(SVD_food.explained_variance_ratio_)
N_food = int(np.argmax(var_acum_food >= 0.90) + 1)

print("Componentes que explican al menos el 90%:", N_food)
print("Varianza acumulada:", round(var_acum_food[N_food-1], 4))
```

```python
# Recomendaciones por correlación de Pearson para Restaurante Pueblo Bonito
SVD_food = TruncatedSVD(n_components=N_food, random_state=1)
resultant_matrix_food = SVD_food.fit_transform(UtMx_food)

corr_mat_food = np.corrcoef(resultant_matrix_food)
restaurante_de_referencia = "Restaurante Pueblo Bonito"
nombres_rest = list(UtMx_food.index)
idx_rest = nombres_rest.index(restaurante_de_referencia)
corr_rest = corr_mat_food[idx_rest]

# Excluimos el restaurante de referencia y conservamos correlaciones positivas
mejores_sim_food = [
    (float(corr_rest[i]), nombres_rest[i])
    for i in range(len(nombres_rest))
    if i != idx_rest and np.isfinite(corr_rest[i]) and corr_rest[i] > 0
]
print("Total de similaridades positivas encontradas:", len(mejores_sim_food))
mejores_sim_food[:5]
```

## **Ejercicio - 5**

```python
# Ordenamos de mayor a menor y mostramos las 10 mejores recomendaciones
mejores_sim_food_ordenadas = sorted(
    mejores_sim_food, key=lambda simil: simil[0], reverse=True
)

print("10 mejores similitudes por evaluación de comida:")
for k, (correlacion, restaurante) in enumerate(mejores_sim_food_ordenadas[:10], start=1):
    print(f"{k}. {restaurante}: {correlacion:.4f}")
```

## **Ejercicio - 6**

### **De manera análoga y usando ahora la evaluación de un restaurante con respecto a su servicio (service_rating), encuentra ahora los diez restaurantes con mayor similaridad al mismo restaurante anterior llamado "Restaurante Pueblo Bonito" y con respecto también a la misma métrica de similaridad de Pearson.**

### **Deberás usar la cantidad de componentes necesarias para que la varabilidad de la factorización SVD quede explicada en un 90%.**

### **a.  Para ello define primero la matriz de utilidad, UtMx_service, donde los renglones son los nombres de los restaurantes, las columnas los usuarios y las entradas las evaluaciones con respecto al servicio recibido en el restaurante.**





```python
# Matriz de utilidad para las evaluaciones del servicio
UtMx_service = df_combinado.pivot_table(
    values="service_rating", index="name", columns="userID", fill_value=0
)

print("Dimensión de la matriz de Utilidad sobre la evaluación del servicio:")
print("(restaurantes, usuarios) =", UtMx_service.shape)
UtMx_service.head()
```

```python
# SVD completa para determinar el máximo número de componentes
U_service, s_service, Vt_service = np.linalg.svd(UtMx_service, full_matrices=False)
nc_service = min(UtMx_service.shape)

SVD_service_completa = TruncatedSVD(n_components=nc_service, random_state=1)
SVD_service_completa.fit(UtMx_service)
var_acum_service = np.cumsum(SVD_service_completa.explained_variance_ratio_)
N_service = int(np.argmax(var_acum_service >= 0.90) + 1)

print("Componentes que explican al menos el 90% del servicio:", N_service)
print("Varianza acumulada:", round(var_acum_service[N_service-1], 4))

# Aplicamos la SVD truncada con ese número de componentes
SVD_service = TruncatedSVD(n_components=N_service, random_state=1)
resultant_matrix_service = SVD_service.fit_transform(UtMx_service)

# Correlación de Pearson contra Restaurante Pueblo Bonito
corr_mat_service = np.corrcoef(resultant_matrix_service)
nombres_service = list(UtMx_service.index)
idx_service = nombres_service.index("Restaurante Pueblo Bonito")
corr_service = corr_mat_service[idx_service]

mejores_sim_service = [
    (float(corr_service[i]), nombres_service[i])
    for i in range(len(nombres_service))
    if i != idx_service and np.isfinite(corr_service[i]) and corr_service[i] > 0
]
mejores_sim_service_ordenadas = sorted(
    mejores_sim_service, key=lambda simil: simil[0], reverse=True
)

print("10 mejores similitudes por evaluación de servicio:")
for k, (correlacion, restaurante) in enumerate(mejores_sim_service_ordenadas[:10], start=1):
    print(f"{k}. {restaurante}: {correlacion:.4f}")
```
