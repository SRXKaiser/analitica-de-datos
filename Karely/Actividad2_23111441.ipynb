5. Emplea boxplots para mostrar la distribución de los datos a través de sus cuartiles.

````python
# Identificar columnas numéricas para graficar sus diagramas de caja
numerical_cols = cr_df.select_dtypes(include=['int64', 'float64']).columns

# Crear subplots para visualizar cada variable numérica de forma individual
plt.figure(figsize=(15, 10))
for i, col in enumerate(numerical_cols, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=cr_df[col], color='skyblue')
    plt.title(f'Boxplot de {col}')
    plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
````

*   Como podrás observar hay valores atípicos en todas las variables. Ejecuta el siguiente código para identificar los valores atípicos en la variable `person_age`

**Interpretación Resultados**

<FONT COLOR="ORANGE">AGREGA AQUI LA INTERPRETACION DE TUS RESULTADOS</FONT>

````python
#COMENTE CADA LINEA DE CODIGO EXPLICANDO SU USO

#.quantile permite calcular los cuartiles faltantes (25% y 75%)
#pasandoles de argumento lo que se indica
#se guarda en una vriable el cuartil de 25% de la categoria person_age
percentile_25 = cr_df["person_age"].quantile(0.25)
#se guarda en una vriable el cuartil de 75% de la categoria person_age
percentile_75 = cr_df["person_age"].quantile(0.75)

#iqr significa rango intercuartilico, osea la medida de dispersion de 50%
#esta se calcula restando el cuartil 75% menos el de 25%
#se guarda el resultando en la variable iqr
iqr = percentile_75 - percentile_25

#se calcula el limite superior con su formula correspondiente y se guarda en una variable
upper_limit = percentile_75 + 1.5 * iqr
#se calcula el limite inferior con su formula correspondiente y se guarda en una variable
lower_limit = percentile_25 - 1.5 * iqr

#se grafica la tabla creando la variable IQR_outliners
#en esta se muestran los datos correspondientes a las personas con edad
#que esten por debajo del limite inferior y encima del limite superior
#no los que estan entre estos dos valores
IQR_outliers = cr_df[(cr_df["person_age"] < lower_limit) | (cr_df["person_age"] > upper_limit)]
#se llama a esta variable para mostrar la tabla
IQR_outliers
````

10. Un mapa de calor con los valores de correlación de todas las variables del dataframe.
*   ¿Qué variable tiene mayor correlación con `loan_status`?

````python
# Calcular la matriz de correlación numérica
corr_matrix = cr_df.corr(numeric_only=True)

# Generar el mapa de calor
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Mapa de Calor de Correlación de Variables Numéricas')
plt.show()
````
